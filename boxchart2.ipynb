```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df2 = pd.read_csv(
    r"C:\Users\Lenovo\Downloads\archive (3)\Sample - Superstore.csv",
    encoding="latin1"
)
```

```python
# Create box plot
plt.figure(figsize=(9, 6))

sns.boxenplot(
    x="Category",
    y="Profit",
    data=df2,
    hue="Category",
    palette="Set1", 
)

plt.title("Profit Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.tight_layout()
plt.show()
```
