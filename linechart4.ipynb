```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load dataset
df4= pd.read_csv(
    r"C:\Users\Lenovo\Downloads\archive (3)\Sample - Superstore.csv",
    encoding="latin1"
)
```

```python
# Convert Order Date to datetime
df4["Order Date"] = pd.to_datetime(df4["Order Date"])

# Calculate yearly sales
yearly_sales = df4.groupby(df4["Order Date"].dt.year)["Sales"].sum()

plt.figure(figsize=(9, 6))
plt.plot(yearly_sales.index, yearly_sales.values,
         marker="o", linewidth=2, color="red",linestyle="--")

plt.title("Sales Trend Over Time")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
