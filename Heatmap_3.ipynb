```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load dataset
df3 = pd.read_csv(
    r"C:\Users\Lenovo\Downloads\archive (3)\Sample - Superstore.csv",
    encoding="latin1"
)
```

```python
# Create pivot table
heatmap_data = pd.pivot_table(
    df3,
    values="Sales",
    index="Region",
    columns="Category",
    aggfunc="sum"
)

# Plot heat map
plt.figure(figsize=(9, 6))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".0f",
    cmap="YlGnBu"
)

plt.title("Heat Map of Sales by Region and Category")
plt.xlabel("Category")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
```
