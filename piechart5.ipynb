```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df5 = pd.read_csv(
    r"C:\Users\Lenovo\Downloads\archive (3)\Sample - Superstore.csv",
    encoding="latin1"
)
```

```python
segment_sales = df5.groupby("Segment")["Sales"].sum()

plt.figure(figsize=(9, 6))

plt.pie(
    segment_sales.values,
    labels=segment_sales.index,
    autopct="%1.1f%%",
    startangle=90,
    explode=(0.02, 0.01, 0.0),
    colors=['#ff9999', '#66b3ff', '#99ff99']
)

plt.title("Sales Distribution by Customer Segment")
plt.show()
```
