```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load dataset
df1 = pd.read_csv(
    r"C:\Users\Lenovo\Downloads\archive (3)\Sample - Superstore.csv",
    encoding="latin1"
)
```

```python
# Sales by Category
category_sales = df1.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 6))
category_sales.plot(kind="bar", color=["lightblue","lightblue", "lightblue"])

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
